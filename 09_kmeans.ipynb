**Dataset:** `datasets/sample_titanic.csv` is an included synthetic teaching dataset, not the official Titanic passenger dataset. Scores are illustrative.

```python
# Load the included sample dataset. No internet or manual input is needed.
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

possible_paths = [
    Path("sample_titanic.csv"),
    Path("../datasets/sample_titanic.csv"),
    Path("datasets/sample_titanic.csv"),
]
csv_file = next((p for p in possible_paths if p.exists()), None)
if csv_file is None:
    raise FileNotFoundError(
        "sample_titanic.csv not found. Keep the datasets folder beside notebooks."
    )

df = pd.read_csv(csv_file)
df = df.rename(columns={
    "PassengerId": "passenger_id", "Survived": "survived", "Pclass": "pclass",
    "Name": "name", "Sex": "sex", "Age": "age", "SibSp": "sibsp",
    "Parch": "parch", "Fare": "fare", "Embarked": "embarked", "Cabin": "cabin"
})
print("Loaded sample dataset:", csv_file)
print("Dataset shape:", df.shape)
display(df.head())
```

## Aim
Cluster passengers into three groups using K-Means.

## 1. Select, clean and scale numeric features

```python
features=["pclass","age","sibsp","parch","fare"]
X=df[features].copy()
for col in features: X[col]=X[col].fillna(X[col].median())
from sklearn.preprocessing import StandardScaler
X_scaled=StandardScaler().fit_transform(X)
print("Dataset shape:",X.shape)
print("Missing values:\n",X.isnull().sum())
```

## 2. Fit K-Means and inspect clusters

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
k=3
model=KMeans(n_clusters=k,random_state=42,n_init=10)
labels=model.fit_predict(X_scaled)
df["KMeans_Cluster"]=labels
print("Cluster centres (standardized feature space):\n",np.round(model.cluster_centers_,3))
print("\nCluster counts:\n",df["KMeans_Cluster"].value_counts().sort_index())
print("\nInertia:",round(model.inertia_,3))
print("Silhouette score:",round(silhouette_score(X_scaled,labels),4))
```

## 3. Plot clusters

```python
plt.figure(figsize=(8,5))
plt.scatter(X["age"],X["fare"],c=labels,cmap="viridis",s=25)
plt.xlabel("Age"); plt.ylabel("Fare"); plt.title("K-Means Clustering (K=3)")
plt.colorbar(label="Cluster"); plt.show()
```

## Expected output
Cluster centres, counts, inertia, silhouette score and scatter plot. Cluster IDs are arbitrary.
