**Dataset:** `datasets/sample_titanic.csv` is an included synthetic teaching dataset, not the official Titanic passenger dataset. Scores are illustrative.

```python
# Load the included sample dataset. No internet or manual input is needed.
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

possible_paths = [
    Path("sample_titanic.csv"),
    Path("../datasets/sample_titanic.csv"),
    Path("datasets/sample_titanic.csv"),
]
csv_file = next((p for p in possible_paths if p.exists()), None)
if csv_file is None:
    raise FileNotFoundError(
        "sample_titanic.csv not found. Keep the datasets folder beside notebooks."
    )

df = pd.read_csv(csv_file)
df = df.rename(columns={
    "PassengerId": "passenger_id", "Survived": "survived", "Pclass": "pclass",
    "Name": "name", "Sex": "sex", "Age": "age", "SibSp": "sibsp",
    "Parch": "parch", "Fare": "fare", "Embarked": "embarked", "Cabin": "cabin"
})
print("Loaded sample dataset:", csv_file)
print("Dataset shape:", df.shape)
display(df.head())
```

## Aim
Inspect the Titanic data, check missing values, clean selected columns, encode categories and visualize the data.

## 1. Inspect dataset

```python
print("\n--- Dataset information ---")
df.info()
print("\n--- Summary statistics ---")
display(df.describe(include="all").T)
print("\n--- Missing values per column ---")
print(df.isnull().sum().sort_values(ascending=False))
```

## 2. Handle missing values and encode categories

```python
clean_df = df.copy()
clean_df["age"] = clean_df["age"].fillna(clean_df["age"].median())
clean_df["embarked"] = clean_df["embarked"].fillna(clean_df["embarked"].mode()[0])
clean_df = clean_df.drop(columns=["deck", "embark_town", "alive", "class", "who", "adult_male", "alone"], errors="ignore")
clean_df = pd.get_dummies(clean_df, columns=["sex", "embarked"], drop_first=True, dtype=int)
print("Missing values after cleaning (largest counts first):")
print(clean_df.isnull().sum().sort_values(ascending=False).head(10))
display(clean_df.head())
```

## 3. Visualizations
Expected: survival count, age distribution, and numeric correlation heatmap.

```python
fig, axes = plt.subplots(1, 3, figsize=(17, 4))
sns.countplot(data=df, x="survived", ax=axes[0])
axes[0].set_title("Survival Count (0=No, 1=Yes)")
sns.histplot(data=df, x="age", bins=20, kde=True, ax=axes[1])
axes[1].set_title("Age Distribution")
sns.heatmap(clean_df.select_dtypes(include=np.number).corr(), cmap="coolwarm", center=0, ax=axes[2])
axes[2].set_title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
